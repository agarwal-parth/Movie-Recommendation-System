```python
# Now the data needs to be cleaned like in the genres I want the Action, Adventure and Science Fiction these labeles for all the movies and the id is not important.
import ast
# def processing_genres(genres):
#     result = []
#     for item in genres:
#         genre = []
#         for i in ast.literal_eval(item):
#             genre.append(i["name"])
#         result.append(genre)
#     return result

def convert(genres):
    result = []
    for item in ast.literal_eval(genres):
        result.append(item['name'])
    return result
```

```python
# def processing_cast(cast):
#     result = []
#     for item in cast:
#         count = 0
#         top_three = []
#         for i in ast.literal_eval(item):
#             if count == 3:
#                 break
#             count += 1
#             top_three.append(i["character"])
#         result.append(top_three)
#     return result

def processing_cast(cast):
    result = []
    count = 0
    for item in ast.literal_eval(cast):
        if count == 3:
            break
        count += 1
        result.append(item["name"])
    return result
```

```python
def fetch_director(crew):
    result = []
    for item in ast.literal_eval(crew):
        if item['job'] == "Director":
            result.append(item["name"])
            break
    return result
```

```python
movies['crew'] = movies['crew'].apply(fetch_director)
```

```python
# Converting the overview into list too so we can easily append

movies['overview'] = movies['overview'].apply(lambda x : x.split())
```

```python
movies.head()
```

```python
# Need to remove the spaces between the words so that extra tags are not being made if someones name is Parth Agarwal and other one is Parth Sharma then the model will create tags for Parth, Agarwal, Parth, Sharma now if someone wants to see the movies for Parth Agarwal then the model will also show the movies of Parth Sharma due to the same tag so we remove the space and are tags are noe ParthAgarwal and ParthSharma.

movies['genres'] = movies['genres'].apply(lambda x : [i.replace(" ", "")for i in x])
```

```python
movies['cast'] = movies['cast'].apply(lambda x : [i.replace(" ", "")for i in x])
```

```python
movies['crew'] = movies['crew'].apply(lambda x : [i.replace(" ", "")for i in x])
```

```python
movies.head()
```

```python
movies['keywords'] = movies['keywords'].apply(lambda x : [i.replace(" ", "")for i in x])
```

```python
movies.head()
```

```python
movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']
```

```python
new_df = movies[['movie_id', 'title', 'tags']]
```

```python
new_df
```

```python
# Used to convert the list into a string.
new_df['tags'] = new_df['tags'].apply(lambda x : " ".join(x))
```

```python
new_df.head()
```

```python
new_df['tags'][0]
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x : x.lower())
```

```python
import nltk
```

```python
# We need to make sure that dance, dancing, danced are all taken as a single feature and so for that using the nltk library.
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    result = []
    for i in text.split():
        result.append(ps.stem(i))
    return " ".join(result)
```

```python
new_df['tags'] = new_df['tags'].apply(stem)
```

```python
# Now doing the vectorization to convert the tags into vectors so they can be compared.
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000, stop_words='english')
```

```python
vectors = cv.fit_transform(new_df['tags']).toarray()
```
